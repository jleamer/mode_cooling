```python
import numpy as np
import qutip as qu
```

```python
σ10 = qu.Qobj([[0, 1, 0],
               [0, 0, 0],
               [0, 0, 0]])
σ21 = qu.Qobj([[0, 0, 0],
               [0, 0, 1],
               [0, 0, 0]])
```

```python
σ10.dag() * σ10 * σ21 - σ21 * σ10.dag() * σ10
```

```python
σ21.dag() * σ21 * σ21 - σ21 * σ21.dag() * σ21
```

```python
σ21 * σ21.dag() - σ21.dag() * σ21
```
